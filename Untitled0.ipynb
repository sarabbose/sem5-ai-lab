```python
from collections import deque
import matplotlib.pyplot as plt
import networkx as nx

# 1. Storing the graph using an adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

# 2. BFS Traversal
def bfs(graph, start):
    visited = []
    queue = deque([start])

    while queue:
        node = queue.popleft()
        if node not in visited:
            visited.append(node)
            queue.extend(graph[node])

    return visited

# 3. DFS Traversal
def dfs(graph, start):
    visited = []

    def dfs_recursive(node):
        if node not in visited:
            visited.append(node)
            for neighbor in graph[node]:
                dfs_recursive(neighbor)

    dfs_recursive(start)
    return visited

# Run Traversals
bfs_order = bfs(graph, 'A')
dfs_order = dfs(graph, 'A')

print("BFS Traversal Order:", bfs_order)
print("DFS Traversal Order:", dfs_order)

# 4. Drawing Visualization with Matplotlib & NetworkX
pos = {
    'A': (0, 2),
    'B': (-1, 1),
    'C': (1, 1),
    'D': (-1.5, 0),
    'E': (-0.5, 0),
    'F': (1, 0)
}

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Create NetworkX Graph
G = nx.DiGraph()
for node, neighbors in graph.items():
    for neighbor in neighbors:
        G.add_edge(node, neighbor)

# Helper function to plot graph with visitation labels
def draw_traversal_tree(ax, traversal_order, title):
    # Map each node to its visit order label (e.g., "A\n(#1)")
    labels = {node: f"{node}\n(#{i+1})" for i, node in enumerate(traversal_order)}

    nx.draw_networkx_nodes(G, pos, ax=ax, node_size=1800, node_color='lightblue', edgecolors='black')
    nx.draw_networkx_edges(G, pos, ax=ax, arrows=True, arrowsize=20, edge_color='gray')
    nx.draw_networkx_labels(G, pos, labels=labels, ax=ax, font_size=10, font_weight='bold')

    ax.set_title(title, fontsize=14, fontweight='bold')
    ax.axis('off')

# Plot BFS
draw_traversal_tree(axes[0], bfs_order, "BFS Traversal Visit Order")

# Plot DFS
draw_traversal_tree(axes[1], dfs_order, "DFS Traversal Visit Order")

plt.tight_layout()
plt.show()
```

```python
from collections import deque

def solve_water_jug():
    cap_a, cap_b = 4, 3
    target = 2

    queue = deque([(0, 0, [])])
    visited = set()

    while queue:
        a, b, path = queue.popleft()

        if a == target or b == target:
            return path + [(a, b)]

        if (a, b) in visited:
            continue
        visited.add((a, b))

        pour_a_to_b = min(a, cap_b - b)
        pour_b_to_a = min(b, cap_a - a)

        current_path = path + [(a, b)]
        states = [
            (cap_a, b),                          # Fill A
            (a, cap_b),                          # Fill B
            (0, b),                              # Empty A
            (a, 0),                              # Empty B
            (a - pour_a_to_b, b + pour_a_to_b),  # Pour A into B
            (a + pour_b_to_a, b - pour_b_to_a)   # Pour B into A
        ]

        for state_a, state_b in states:
            if (state_a, state_b) not in visited:
                queue.append((state_a, state_b, current_path))

    return None

solution = solve_water_jug()
if solution:
    print("Minimum sequence of steps to measure 2 Litres:\n(Jug A [4L], Jug B [3L])")
    for step, state in enumerate(solution):
        print(f"Step {step}: {state}")
```
